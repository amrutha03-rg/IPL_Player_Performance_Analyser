```python
import pandas as pd
```

```python
matches = pd.read_csv("/content/combined_matches_cleaned.csv")
deliveries = pd.read_csv("/content/combined_deliveries_cleaned.csv")
print("Datasets loaded successfully!")
```

```python
top_scorers = deliveries.groupby("batter")["runs_batter"].sum()
top_scorers = top_scorers.sort_values(ascending=False).head(10)
display(top_scorers.reset_index())
```

```python
top_wickets = deliveries.groupby("bowler")["wicket"].sum()
top_wickets = top_wickets.sort_values(ascending=False).head(10)
display(top_wickets.reset_index())
```

```python
teams = pd.concat([
    matches["team1"],
    matches["team2"]
]).dropna().unique()
team_analysis = pd.DataFrame({"Team": teams})
team_analysis["Matches"] = team_analysis["Team"].apply(
    lambda t: ((matches["team1"] == t) | (matches["team2"] == t)).sum())
team_analysis["Wins"] = team_analysis["Team"].apply(lambda t: (matches["winner"] == t).sum())
team_analysis["Win_Percentage"] = (team_analysis["Wins"] /team_analysis["Matches"] * 100).round(2)
display(team_analysis.sort_values("Win_Percentage", ascending=False))
```

```python
season_analysis = matches.groupby("season")["match_id"].nunique().reset_index()
season_analysis.columns = ["Season", "Matches"]
display(season_analysis)
```

```python
player_runs = deliveries.groupby("batter")["runs_batter"].sum()
player_outs = deliveries[
    deliveries["wicket"] == 1
].groupby("batter").size()
batting_average = pd.DataFrame({
    "Total_Runs": player_runs,
    "Outs": player_outs}).fillna(0)
batting_average["Batting_Average"] = (
    batting_average["Total_Runs"] /
    batting_average["Outs"].replace(0, pd.NA))
display(batting_average.sort_values("Batting_Average", ascending=False).head(10))
```

```python
toss_wins = ( matches["toss_winner"] == matches["winner"]).sum()
total_matches = matches.dropna(subset=["toss_winner", "winner"]).shape[0]
toss_percentage = round(toss_wins / total_matches * 100, 2)
print("Matches with toss and winner data:", total_matches)
print("Toss winner also won:", toss_wins)
print("Percentage:", toss_percentage, "%")
```
