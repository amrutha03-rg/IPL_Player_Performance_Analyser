```python
import pandas as pd
```

```python
batting = pd.read_csv("/content/batting_summary (1).csv")
bowling = pd.read_csv("/content/bowling_summary (1).csv")
team = pd.read_csv("/content/team_summary (1).csv")
season = pd.read_csv("/content/season_summary (1).csv")
print("All files loaded successfully!")
```

```python
print("BATTING RESULTS")
display(batting.head())
print("BOWLING RESULTS")
display(bowling.head())
print("TEAM RESULTS")
display(team.head())
print("SEASON RESULTS")
display(season.head())
```

```python
top_batter = batting.nlargest(1, "Total_Runs").iloc[0]
print("Top Batter:", top_batter["batter"])
print("Total Runs:", top_batter["Total_Runs"])
```

```python
top_bowler = bowling.nlargest(1, "Wickets").iloc[0]
print("Top Bowler:", top_bowler["bowler"])
print("Wickets:", top_bowler["Wickets"])
```

```python
top_team = team.nlargest(1, "Win_Percentage").iloc[0]
print("Top Team:", top_team["Team"])
print("Win Percentage:",
      round(top_team["Win_Percentage"], 2))
```

```python
most_matches = season.nlargest(1, "Total_Matches").iloc[0]
print("Season:", most_matches["season"])
print("Matches:", most_matches["Total_Matches"])
```

```python
print("RESULTS & INTERPRETATION")
print("Batting data identifies the highest run scorer.")
print("Bowling data identifies the highest wicket taker.")
print("Team data compares win percentages.")
print("Season data shows match-count trends.")
```

```python
print("CONCLUSION")
print("IPL data was analyzed across four performance areas:")
print("batting, bowling, teams, and seasons.")
print("The summaries and visualizations help identify")
print("historical patterns and compare performance.")
```
