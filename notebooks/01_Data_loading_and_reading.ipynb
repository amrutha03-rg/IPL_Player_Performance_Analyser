## DATASET-1 (2008-2019)

```python
from zipfile import ZipFile

with ZipFile('2008-2019.zip', 'r') as zip_ref:
    zip_ref.extractall('ipl_dataset-1')
```

```python
import os

os.listdir("ipl_dataset-1")
```

```python
#loading the files in dataset-1
import pandas as pd

matches_df = pd.read_csv("ipl_dataset-1/matches.csv")
deliveries_df = pd.read_csv("ipl_dataset-1/deliveries.csv")

print("Both datasets loaded successfully!")
```

```python
#size of the files in dataset-1
print("Matches dataset shape:", matches_df.shape)
print("Deliveries dataset shape:", deliveries_df.shape)
```

```python
#First five rows of each file in dataset-1
print("Matches dataset:")
display(matches_df.head())

print("\nDeliveries dataset:")
display(deliveries_df.head())
```

```python
#seasons covered in dataset-1
print("Seasons in Dataset 1:")
print(matches_df["Season"].unique())
```

```python
#columns present in matches and deliveries csv files in dataset-1
print("Matches columns:")
print(matches_df.columns.tolist())

print("\nDeliveries columns:")
print(deliveries_df.columns.tolist())
```

```python
#checking datatypes
print("Matches dataset data types:")
print(matches_df.dtypes)

print("\nDeliveries dataset data types:")
print(deliveries_df.dtypes)
```

observed that date is not in proper form (stored as text) and will be changed later

```python
#Check the unique values of important categorical columns
print("Toss decisions:")
print(matches_df["toss_decision"].unique())

print("\nMatch results:")
print(matches_df["result"].unique())

print("\nDismissal types:")
print(deliveries_df["dismissal_kind"].dropna().unique())
```

Dataset-1 (2008-2019) is checked, analysed & intial cleaning is done.

```python
import os

print(os.listdir("ipl_dataset_2"))
```

```python
import glob

csv_files = glob.glob("ipl_dataset_2/*.csv")

print("CSV files found:")
print(csv_files)
```

```python
#loading files in dataset using glob
import pandas as pd

deliveries_df2 = pd.read_csv(csv_files[0])
matches_df2 = pd.read_csv(csv_files[1])

print("Dataset 2 files loaded successfully!")
print("Deliveries shape:", deliveries_df2.shape)
print("Matches shape:", matches_df2.shape)
```

```python
print("Matches columns:")
print(matches_df2.columns.tolist())

print("\\nDeliveries columns:")
print(deliveries_df2.columns.tolist())

print("\\nSeasons available:")
print(sorted(matches_df2["season"].unique(), key=str))
```
