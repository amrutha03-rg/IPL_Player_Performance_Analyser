```python
# Import Pandas
import pandas as pd

# Load Dataset 1 extracted files from Stage 3
matches_df1 = pd.read_csv("matches_df1_extracted.csv")
deliveries_df1 = pd.read_csv("deliveries_df1_extracted.csv")

# Check whether the files were loaded successfully
print("Dataset 1 extracted data loaded successfully!")

# Display the dimensions of both datasets
print("Matches shape:", matches_df1.shape)
print("Deliveries shape:", deliveries_df1.shape)
```

```python
# Check missing values in Dataset 1 matches
print("Missing values in Dataset 1 matches:")
print(matches_df1.isnull().sum())

print("\n" + "=" * 60)

# Check missing values in Dataset 1 deliveries
print("Missing values in Dataset 1 deliveries:")
print(deliveries_df1.isnull().sum())
```

```python
# Check for completely identical duplicate rows

print("Duplicate match rows:", matches_df1.duplicated().sum())
print("Duplicate delivery rows:", deliveries_df1.duplicated().sum())
```

```python
# Inspect all delivery rows that belong to exact duplicate groups
# keep=False shows every row involved in a duplicate group

duplicate_deliveries = deliveries_df1[
    deliveries_df1.duplicated(keep=False)
].sort_values(
    by=["match_id", "inning", "over", "ball"]
)

print("Number of rows involved in exact duplicate groups:",
      len(duplicate_deliveries))

print("\nDuplicate delivery records:")
display(duplicate_deliveries)
```

```python
# Remove only completely identical delivery rows
# No subset of columns is used, so legitimate repeated delivery positions are preserved

deliveries_df1 = deliveries_df1.drop_duplicates().copy()

print("Dataset 1 deliveries after removing exact duplicates:")
print("Shape:", deliveries_df1.shape)
print("Remaining exact duplicate rows:", deliveries_df1.duplicated().sum())
```

```python
# Final verification after removing exact duplicate rows

print("Duplicate delivery rows after cleaning:",
      deliveries_df1.duplicated().sum())

print("Current Dataset 1 deliveries shape:",
      deliveries_df1.shape)
```

```python
# Inspect the data types of Dataset 1 matches
print("Dataset 1 matches datatypes:")
print(matches_df1.dtypes)

print("\n" + "=" * 60)

# Inspect the data types of Dataset 1 deliveries
print("Dataset 1 deliveries datatypes:")
print(deliveries_df1.dtypes)
```

```python
# Inspect all unique Season values before standardization

print("Unique Season values in Dataset 1:")
print(sorted(matches_df1["Season"].unique()))
```

```python
# Validate whether every Season value follows the expected IPL-YYYY format

season_format_check = matches_df1["Season"].str.match(r"^IPL-\d{4}$")

print("All Season values follow IPL-YYYY format:",
      season_format_check.all())

print("Number of invalid Season values:",
      (~season_format_check).sum())
```

```python
# Standardize Season values by removing the "IPL-" prefix
# and converting the remaining year to integer

matches_df1["Season"] = (
    matches_df1["Season"]
    .str.replace("IPL-", "", regex=False)
    .astype(int)
)

print("Dataset 1 Season values after standardization:")
print(sorted(matches_df1["Season"].unique()))

print("\nSeason datatype:")
print(matches_df1["Season"].dtype)
```

```python
# Inspect the current date format before conversion

print("Dataset 1 date datatype:")
print(matches_df1["date"].dtype)

print("\nFirst 10 date values:")
print(matches_df1["date"].head(10).to_list())

print("\nNumber of unique date values:")
print(matches_df1["date"].nunique())
```

```python
# Keep only the date part and remove the time component
matches_df1["date"] = matches_df1["date"].dt.date

print("Dataset 1 date datatype after removing time:")
print(matches_df1["date"].dtype)

print("\nFirst 10 dates:")
print(matches_df1["date"].head(10).to_list())
```

```python
# Inspect important categorical columns before standardization

categorical_columns = [
    "team1",
    "team2",
    "toss_winner",
    "toss_decision",
    "winner",
    "result",
    "player_of_match"
]

for column in categorical_columns:
    print("\n" + "=" * 60)
    print("Unique values in", column, ":")
    print(matches_df1[column].dropna().unique())
    print("Number of unique values:", matches_df1[column].nunique())
```

```python
# Standardize Dataset 1 match column names

matches_df1 = matches_df1.rename(columns={
    "id": "match_id",
    "Season": "season",
    "result": "result_type"
})

print("Dataset 1 match columns after standardization:")
print(matches_df1.columns.tolist())
```

```python
# Standardize Dataset 1 delivery column names

deliveries_df1 = deliveries_df1.rename(columns={
    "inning": "innings",
    "batsman": "batter",
    "batsman_runs": "runs_batter",
    "extra_runs": "runs_extras",
    "total_runs": "runs_total",
    "player_dismissed": "wicket_player_out",
    "dismissal_kind": "wicket_kind"
})

print("Dataset 1 delivery columns after standardization:")
print(deliveries_df1.columns.tolist())
```

```python
# Load Dataset 2 extracted data

matches_df2 = pd.read_csv("matches_df2_extracted.csv")
deliveries_df2 = pd.read_csv("deliveries_df2_extracted.csv")

print("Dataset 2 extracted data loaded successfully!")
print("Matches shape:", matches_df2.shape)
print("Deliveries shape:", deliveries_df2.shape)
```

```python
# Inspect wicket-related values in Dataset 2

print("Unique values in Dataset 2 wicket_player_out:")
print(deliveries_df2["wicket_player_out"].dropna().unique())

print("\n" + "=" * 60)

print("Unique values in Dataset 2 wicket_kind:")
print(deliveries_df2["wicket_kind"].dropna().unique())
```

```python
# Inspect Dataset 1 match-result columns

print("Dataset 1 result-related columns:")
print(matches_df1[["result_type", "winner"]].head())

print("\nUnique values in result_type:")
print(matches_df1["result_type"].dropna().unique())

print("\nNumber of missing values:")
print(matches_df1[["result_type", "winner"]].isnull().sum())
```

```python
# Inspect remaining Dataset 1 match columns

print("Dataset 1 match columns:")
print(matches_df1.columns.tolist())
```

```python
# Check team names in Dataset 1

team_columns = ["team1", "team2", "toss_winner", "winner"]

for column in team_columns:
    print("\n" + "=" * 60)
    print("Team values in", column, ":")
    print(matches_df1[column].dropna().unique())
```

```python
# Standardize historical team names in Dataset 1

team_name_mapping = {
    "Rising Pune Supergiant": "Rising Pune Supergiants"
}

for column in ["team1", "team2", "toss_winner", "winner"]:
    matches_df1[column] = matches_df1[column].replace(team_name_mapping)

print("Historical team names standardized successfully!")
```

```python
# Verify standardized team names in Dataset 1

for column in ["team1", "team2", "toss_winner", "winner"]:
    print("\n" + "=" * 60)
    print("Team values in", column, ":")
    print(matches_df1[column].dropna().unique())
```

```python
# Check missing values in Dataset 2

print("Missing values in Dataset 2 matches:")
print(matches_df2.isnull().sum())

print("\n" + "=" * 60)

print("Missing values in Dataset 2 deliveries:")
print(deliveries_df2.isnull().sum())
```

```python
# Check exact duplicate records in Dataset 2

print("Duplicate match rows:",
      matches_df2.duplicated().sum())

print("Duplicate delivery rows:",
      deliveries_df2.duplicated().sum())
```

```python
# Inspect Dataset 2 datatypes

print("Dataset 2 matches datatypes:")
print(matches_df2.dtypes)

print("\n" + "=" * 60)

print("Dataset 2 deliveries datatypes:")
print(deliveries_df2.dtypes)
```

```python
# Inspect Dataset 2 date values before conversion

print("Dataset 2 date datatype:")
print(matches_df2["date"].dtype)

print("\nFirst 10 date values:")
print(matches_df2["date"].head(10).to_list())
```

```python
# Convert Dataset 2 date from text to proper date format
# The source data uses YYYY-MM-DD format

matches_df2["date"] = pd.to_datetime(
    matches_df2["date"],
    format="%Y-%m-%d"
).dt.date

print("Dataset 2 date datatype after conversion:")
print(matches_df2["date"].dtype)

print("\nFirst 10 dates:")
print(matches_df2["date"].head(10).to_list())
```

```python
# Check team names in Dataset 2

team_columns = ["team1", "team2", "toss_winner", "winner"]

for column in team_columns:
    print("\n" + "=" * 60)
    print("Team values in", column, ":")
    print(matches_df2[column].dropna().unique())
```

```python
# Inspect Dataset 2 result-related columns

print("Dataset 2 result-related columns:")
print(matches_df2[["result_type", "winner"]].head())

print("\nUnique values in result_type:")
print(matches_df2["result_type"].dropna().unique())

print("\nNumber of missing values:")
print(matches_df2[["result_type", "winner"]].isnull().sum())
```

```python
# Verify Dataset 2 match columns

print("Dataset 2 match columns:")
print(matches_df2.columns.tolist())
```

```python
# Create copies of the cleaned datasets for final standardization and combination

matches_clean1 = matches_df1.copy()
deliveries_clean1 = deliveries_df1.copy()

matches_clean2 = matches_df2.copy()
deliveries_clean2 = deliveries_df2.copy()

print("Working copies created successfully!")
print("Dataset 1 matches:", matches_clean1.shape)
print("Dataset 1 deliveries:", deliveries_clean1.shape)
print("Dataset 2 matches:", matches_clean2.shape)
print("Dataset 2 deliveries:", deliveries_clean2.shape)
```

```python
# Create a common wicket indicator for Dataset 1
# A wicket is recorded when wicket_player_out is not missing

deliveries_clean1["wicket"] = (
    deliveries_clean1["wicket_player_out"].notna().astype(int)
)

print("Dataset 1 wicket indicator created successfully!")
print("\nWicket indicator counts:")
print(deliveries_clean1["wicket"].value_counts().sort_index())
```

```python
# Compare Dataset 1 and Dataset 2 delivery columns

print("Dataset 1 delivery columns:")
print(deliveries_clean1.columns.tolist())

print("\n" + "=" * 60)

print("Dataset 2 delivery columns:")
print(deliveries_clean2.columns.tolist())
```

```python
# Compare Dataset 1 and Dataset 2 match columns

print("Dataset 1 match columns:")
print(matches_clean1.columns.tolist())

print("\n" + "=" * 60)

print("Dataset 2 match columns:")
print(matches_clean2.columns.tolist())
```

```python
# Align Dataset 2 match columns with Dataset 1 before combining

matches_clean2 = matches_clean2[matches_clean1.columns]

print("Dataset 2 match columns after alignment:")
print(matches_clean2.columns.tolist())
```

```python
# Combine cleaned match data from Dataset 1 and Dataset 2

combined_matches = pd.concat(
    [matches_clean1, matches_clean2],
    ignore_index=True
)

print("Combined match data created successfully!")
print("Combined matches shape:", combined_matches.shape)
```

```python
# Validate the combined match data

print("Combined match shape:", combined_matches.shape)

print("\nExact duplicate rows:")
print(combined_matches.duplicated().sum())

print("\nDuplicate match IDs:")
print(combined_matches["match_id"].duplicated().sum())

print("\nUnique match IDs:")
print(combined_matches["match_id"].nunique())
```

```python
# Combine cleaned delivery data from Dataset 1 and Dataset 2

combined_deliveries = pd.concat(
    [deliveries_clean1, deliveries_clean2],
    ignore_index=True
)

print("Combined delivery data created successfully!")
print("Combined deliveries shape:", combined_deliveries.shape)
```

```python
# Validate that every delivery belongs to a valid match

valid_match_ids = set(combined_matches["match_id"])

invalid_delivery_matches = combined_deliveries[
    ~combined_deliveries["match_id"].isin(valid_match_ids)
]

print("Total delivery rows:", len(combined_deliveries))
print("Invalid delivery rows:", len(invalid_delivery_matches))
print("Unique invalid match IDs:",
      invalid_delivery_matches["match_id"].nunique())
```

```python
# Inspect repeated delivery positions
# These are not automatically duplicate records.

delivery_key = ["match_id", "innings", "over", "ball"]

repeated_delivery_positions = combined_deliveries[
    combined_deliveries.duplicated(
        subset=delivery_key,
        keep=False
    )
].sort_values(by=delivery_key)

print("Rows involved in repeated delivery positions:",
      len(repeated_delivery_positions))

print("Number of repeated delivery positions:",
      repeated_delivery_positions[delivery_key]
      .drop_duplicates()
      .shape[0])

display(repeated_delivery_positions)
```

```python
# Final exact-duplicate check for combined delivery data

print("Exact duplicate delivery rows:",
      combined_deliveries.duplicated().sum())

print("Combined delivery shape:",
      combined_deliveries.shape)
```

```python
# Final missing-value check for combined match data

print("Missing values in combined matches:")
print(combined_matches.isnull().sum())
```

```python
# Inspect matches with missing values in important columns

missing_match_records = combined_matches[
    combined_matches["city"].isnull()
    | combined_matches["winner"].isnull()
    | combined_matches["player_of_match"].isnull()
]

print("Number of matches with at least one missing value:",
      len(missing_match_records))

display(
    missing_match_records[
        [
            "match_id",
            "season",
            "date",
            "city",
            "venue",
            "team1",
            "team2",
            "winner",
            "result_type",
            "player_of_match"
        ]
    ]
)
```

```python
# Analyze missing winner and player_of_match values by result type

missing_by_result = combined_matches.groupby("result_type").agg(
    total_matches=("match_id", "count"),
    missing_winner=("winner", lambda x: x.isnull().sum()),
    missing_player_of_match=("player_of_match", lambda x: x.isnull().sum())
)

print("Missing winner and player_of_match values by result type:")
display(missing_by_result)
```

```python
# Final missing-value check for combined delivery data

print("Missing values in combined deliveries:")
print(combined_deliveries.isnull().sum())
```

```python
# Standardize historical team names

team_name_map = {
    "Delhi Daredevils": "Delhi Capitals",
    "Kings XI Punjab": "Punjab Kings",
    "Royal Challengers Bangalore": "Royal Challengers Bengaluru",
    "Rising Pune Supergiant": "Rising Pune Supergiants"
}

# Match dataset
for column in ["team1", "team2", "toss_winner", "winner"]:
    combined_matches[column] = combined_matches[column].replace(
        team_name_map
    )

# Delivery dataset
if "batting_team" in combined_deliveries.columns:
    combined_deliveries["batting_team"] = (
        combined_deliveries["batting_team"].replace(team_name_map)
    )

print("Team names standardized successfully!")
```

```python
# Verify standardized team names

print("Teams in combined match data:")

print(
    sorted(
        set(combined_matches["team1"].dropna()) |
        set(combined_matches["team2"].dropna())
    )
)

print("\nTeams in combined delivery data:")

print(
    sorted(
        set(combined_deliveries["batting_team"].dropna())
    )
)
```

```python
# Final structure validation

print("Combined matches shape:", combined_matches.shape)
print("Combined matches columns:")
print(list(combined_matches.columns))

print("\n" + "=" * 60)

print("Combined deliveries shape:", combined_deliveries.shape)
print("Combined deliveries columns:")
print(list(combined_deliveries.columns))
```

```python
# Final validation: check that every match has delivery records

match_ids = set(combined_matches["match_id"])
delivery_match_ids = set(combined_deliveries["match_id"])

matches_without_deliveries = match_ids - delivery_match_ids
deliveries_without_matches = delivery_match_ids - match_ids

print("Total matches:", len(match_ids))
print("Matches without deliveries:", len(matches_without_deliveries))
print("Delivery match IDs without match records:",
      len(deliveries_without_matches))
```

```python
# Save final cleaned and standardized datasets

combined_matches.to_csv(
    "combined_matches_cleaned.csv",
    index=False
)

combined_deliveries.to_csv(
    "combined_deliveries_cleaned.csv",
    index=False
)

print("Final cleaned datasets saved successfully!")
print("Saved: combined_matches_cleaned.csv")
print("Saved: combined_deliveries_cleaned.csv")
```
