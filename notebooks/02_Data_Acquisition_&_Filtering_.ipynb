```python
import zipfile

with zipfile.ZipFile("2008-26.zip", "r") as zip_ref:
    zip_ref.extractall("ipl_dataset_2")

print("Dataset 2 extracted successfully!")
```

```python
import os

print(os.listdir("ipl_dataset_2"))
```

```python
import glob

csv_files = glob.glob("ipl_dataset_2/*.csv")

print("CSV files found:")
print(csv_files)
```

```python
import pandas as pd

matches_df2 = pd.read_csv("ipl_dataset_2/ipl_matches_clean.csv")
deliveries_df2 = pd.read_csv("ipl_dataset_2/ipl_deliveries_clean.csv")

print("Dataset 2 loaded successfully!")
print("Matches shape:", matches_df2.shape)
print("Deliveries shape:", deliveries_df2.shape)
```

```python
#Check available seasons
print("Seasons available in Dataset 2:")
print(sorted(matches_df2["season"].unique()))
```

```python
#Filter matches for 2020–2026
required_seasons = list(range(2020, 2027))

matches_2020_2026 = matches_df2[
    matches_df2["season"].isin(required_seasons)
].copy()

print("Filtered matches shape:", matches_2020_2026.shape)
print("\nSeasons after filtering:")
print(sorted(matches_2020_2026["season"].unique()))
```

```python
# Filter the corresponding deliveries
required_match_ids = matches_2020_2026["match_id"].unique()

deliveries_2020_2026 = deliveries_df2[
    deliveries_df2["match_id"].isin(required_match_ids)
].copy()

print("Filtered deliveries shape:", deliveries_2020_2026.shape)
print("Unique match IDs in filtered deliveries:",
      deliveries_2020_2026["match_id"].nunique())
```

```python
# Filter corresponding deliveries
required_match_ids = matches_2020_2026["match_id"].unique()

deliveries_2020_2026 = deliveries_df2[
    deliveries_df2["match_id"].isin(required_match_ids)
].copy()

print("Filtered deliveries shape:", deliveries_2020_2026.shape)
print("Unique match IDs in filtered deliveries:",
      deliveries_2020_2026["match_id"].nunique())
```

```python
# Verify the season distribution
season_match_counts = matches_2020_2026["season"].value_counts().sort_index()

print("Matches acquired for each season:")
print(season_match_counts)
```

```python
#Save the filtered datasets
matches_2020_2026.to_csv("matches_2020_2026.csv", index=False)
deliveries_2020_2026.to_csv("deliveries_2020_2026.csv", index=False)

print("Filtered datasets saved successfully!")
print("Matches file: matches_2020_2026.csv")
print("Deliveries file: deliveries_2020_2026.csv")
```
