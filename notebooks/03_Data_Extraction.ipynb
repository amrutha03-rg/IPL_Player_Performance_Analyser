DATASET-1(2008-2019)

```python
from zipfile import ZipFile

with ZipFile('2008-2019.zip', 'r') as zip_ref:
    zip_ref.extractall('ipl_dataset-1')
```

```python
import os

os.listdir("ipl_dataset-1")
```

```python
#loading the files in dataset-1
import pandas as pd

matches_df = pd.read_csv("ipl_dataset-1/matches.csv")
deliveries_df = pd.read_csv("ipl_dataset-1/deliveries.csv")

print("Both datasets loaded successfully!")
```

```python
print('Shape of matches',matches_df.shape)
print("Columns",matches_df.columns)
```

```python
print('Shape of deliveries',deliveries_df.shape)
print('columns',deliveries_df.columns)
```

```python
# Extract required columns from Dataset 1 match-level data

matches_df1_extracted = matches_df[
    [
        "id",
        "Season",
        "date",
        "city",
        "venue",
        "team1",
        "team2",
        "toss_winner",
        "toss_decision",
        "winner",
        "result",
        "player_of_match"
    ]
]

print("Dataset 1 match data extracted successfully!")
print("Shape:", matches_df1_extracted.shape)
print("Columns:")
print(matches_df1_extracted.columns)
```

```python
# Extract required columns from Dataset 1 delivery-level data

deliveries_df1_extracted = deliveries_df[
    [
        "match_id",
        "inning",
        "batting_team",
        "over",
        "ball",
        "batsman",
        "bowler",
        "non_striker",
        "batsman_runs",
        "extra_runs",
        "total_runs",
        "player_dismissed",
        "dismissal_kind"
    ]
]

print("Dataset 1 delivery data extracted successfully!")
print("Shape:", deliveries_df1_extracted.shape)
print("Columns:")
print(deliveries_df1_extracted.columns)
```

DATASET-2(2020-2026)

```python
import zipfile

with zipfile.ZipFile("2020-26.zip", "r") as zip_ref:
    zip_ref.extractall("ipl_dataset_2")

print("Dataset 2 extracted successfully!")
```

```python
import os

print(os.listdir("ipl_dataset_2"))
```

```python
#loading files in dataset-2
import pandas as pd

matches_df2 = pd.read_csv("ipl_dataset_2/matches_2020_2026.csv")
deliveries_df2 = pd.read_csv("ipl_dataset_2/deliveries_2020_2026.csv")

print("Dataset 2 files loaded successfully!")
print("Deliveries shape:", deliveries_df2.shape)
print("Matches shape:", matches_df2.shape)
```

```python
#columns present in matches and deliveries csv files in dataset-1
print("Matches columns:")
print(matches_df2.columns.tolist())

print("\nDeliveries columns:")
print(deliveries_df2.columns.tolist())
```

```python
# Extract required columns from Dataset 2 match-level data

matches_df2_extracted = matches_df2[
    [
        "match_id",
        "date",
        "season",
        "venue",
        "city",
        "team1",
        "team2",
        "toss_winner",
        "toss_decision",
        "winner",
        "result_type",
        "player_of_match"
    ]
]

print("Dataset 2 match data extracted successfully!")
print("Shape:", matches_df2_extracted.shape)
print("Columns:")
print(matches_df2_extracted.columns.tolist())
```

```python
# Extract required columns from Dataset 2 delivery-level data

deliveries_df2_extracted = deliveries_df2[
    [
        "match_id",
        "innings",
        "batting_team",
        "over",
        "ball",
        "batter",
        "bowler",
        "non_striker",
        "runs_batter",
        "runs_extras",
        "runs_total",
        "wicket_player_out",
        "wicket_kind"
    ]
]

print("Dataset 2 delivery data extracted successfully!")
print("Shape:", deliveries_df2_extracted.shape)
print("Columns:")
print(deliveries_df2_extracted.columns.tolist())
```

```python
# Save the extracted datasets for Stage 4

matches_df1_extracted.to_csv("matches_df1_extracted.csv", index=False)
deliveries_df1_extracted.to_csv("deliveries_df1_extracted.csv", index=False)

matches_df2_extracted.to_csv("matches_df2_extracted.csv", index=False)
deliveries_df2_extracted.to_csv("deliveries_df2_extracted.csv", index=False)

print("All four extracted datasets saved successfully!")
```
