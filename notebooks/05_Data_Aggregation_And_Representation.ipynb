```python
import pandas as pd
```

```python
matches = pd.read_csv("/content/combined_matches_cleaned.csv")
deliveries = pd.read_csv("/content/combined_deliveries_cleaned.csv")
print("Matches:", matches.shape)
print("Deliveries:", deliveries.shape)
```

```python
display(matches.head())
display(deliveries.head())
```

```python
#Batting aggregation
batting = deliveries.groupby("batter").agg(
    Total_Runs=("runs_batter", "sum"),
    Matches=("match_id", "nunique")).reset_index()
batting = batting.sort_values("Total_Runs", ascending=False)
display(batting.head(10))
```

```python
#Bowling aggreagation
bowling = deliveries.groupby("bowler").agg(
    Total_Runs=("runs_total", "sum"),
    Matches=("match_id", "nunique"),
    Wickets=("wicket", "sum")
).reset_index()
bowling = bowling.sort_values("Wickets", ascending=False)
display(bowling.head(10))
```

```python
#Team aggregation
team1 = matches[
    ["match_id", "team1", "winner"]
].rename(columns={"team1": "Team"})
team2 = matches[
    ["match_id", "team2", "winner"]
].rename(columns={"team2": "Team"})
team_data = pd.concat(
    [team1, team2],
    ignore_index=True)
team_data["Win"] = (
    team_data["Team"] == team_data["winner"]
).astype(int)
team_performance = team_data.groupby("Team").agg(
    Matches=("match_id", "nunique"),
    Wins=("Win", "sum")
).reset_index()
team_performance["Win_Percentage"] = (team_performance["Wins"] /team_performance["Matches"] * 100)
display(team_performance.sort_values("Win_Percentage", ascending=False))
```

```python
#season wise aggregation
season_performance = matches.groupby("season").agg(
    Total_Matches=("match_id", "nunique")).reset_index()
display(season_performance)
```

```python
from pathlib import Path
output = Path("outputs")
output.mkdir(exist_ok=True)
batting.to_csv(output / "batting_summary.csv",index=False)
bowling.to_csv(output / "bowling_summary.csv",index=False)
team_performance.to_csv(output / "team_summary.csv",index=False)
season_performance.to_csv(output / "season_summary.csv",index=False)
print("Stage 5 completed successfully!")
```
